```python
import pandas as pd

df25 = pd.read_csv("2025.csv")
df26 = pd.read_csv("2026PartYear.csv")

print("2025 file:", df25.shape)
print("2026 file:", df26.shape)        (CHECKED)
```

```python
periods = df26["Period"].unique()

us25 = df25[
    (df25["Geography"] == "Total U.S.") &
    (df25["Period"].isin(periods))
]

us26 = df26[
    df26["Geography"] == "Total U.S."
]

print("2025 weeks:", us25["Period"].min(), "to", us25["Period"].max())
print("2026 weeks:", us26["Period"].min(), "to", us26["Period"].max())             (CHECKED)
```

```python
volume25 = us25["Total Bulk and Bags"].sum()
volume26 = us26["Total Bulk and Bags"].sum()

volume_change = ((volume26 - volume25) / volume25) * 100

print("2025 Total Volume:", round(volume25))
print("2026 Total Volume:", round(volume26))
print("Volume Change:", round(volume_change, 1), "%")                     (CHECKED)
```

```python
asp25 = (us25["ASP Current Year"] * us25["Total Bulk and Bags"]).sum() / volume25
asp26 = (us26["ASP Current Year"] * us26["Total Bulk and Bags"]).sum() / volume26

asp_change = ((asp26 - asp25) / asp25) * 100

print("2025 Average Price: $", round(asp25, 2))
print("2026 Average Price: $", round(asp26, 2))
print("Price Change:", round(asp_change, 1), "%")                 (CHECKED)
```

```python
for avocado_type in ["Conventional", "Organic"]:

    type25 = us25[us25["Type"] == avocado_type]
    type26 = us26[us26["Type"] == avocado_type]

    vol25 = type25["Total Bulk and Bags"].sum()
    vol26 = type26["Total Bulk and Bags"].sum()

    change = ((vol26 - vol25) / vol25) * 100

    print(avocado_type)
    print("2025 Volume:", round(vol25))
    print("2026 Volume:", round(vol26))
    print("Change:", round(change, 1), "%")
    print()                                                (CHECKED)
```

```python
bags25 = us25["TotalBagged"].sum()
bags26 = us26["TotalBagged"].sum()

bag_change = ((bags26 - bags25) / bags25) * 100

share25 = (bags25 / volume25) * 100
share26 = (bags26 / volume26) * 100

print("Bagged Volume Change:", round(bag_change, 1), "%")
print("2025 Bagged Share:", round(share25, 1), "%")
print("2026 Bagged Share:", round(share26, 1), "%")                         (CHECKED)
```
